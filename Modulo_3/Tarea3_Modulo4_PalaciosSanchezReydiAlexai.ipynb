```python
import numpy as np
from keras.models import Sequential #Así se llama la librería
from keras.layers import Dense # Así se llama la capa
```

```python
# cargamos las 4 combinaciones de las compuertas XOR
training_data = np.array([[0,0 ],[0,1 ],[1,0],[1,1]])

# y estos son los resultados que se obtienen, en el mismo orden
target_data = np.array([[[0 ],[ 0],[ 0]],[[ 1],[0],[1]],[[0],[1],[1]],[[1],[1],[1]]])
```

```python
print(target_data)
```

```python
print(training_data)
```

# Definimos una arquitectura de red neuronal

```python
model = Sequential() #Aquí le asignamos un nombre o variable al modelo Sequential
model.add(Dense(200, input_dim=2, activation='relu')) #add es agrega y dice, agrega una capa densa, el 16 es el número de neuronas, input dim son las dimensiones para este caso son dos, la función de activación ("relu") es como el airling, se le conoce como la función de rectificación
model.add(Dense(3, activation='sigmoid')) #en el primer parámetro le ponemos 1 porque basta una neurona para que nos dé un cero y un 1 y de función de activación se usa la sigmoide
```

```python
#Ahora le vamos a dar los datos de entrenamiento
model.compile(loss='mean_squared_error', #Aquí le ponemos la función para evaluar el error, usaremos la función de costo o de error cuadrático medio
              optimizer='adam', #En optimizador le ponemos adam, mañana lo explica
              metrics=['binary_accuracy'])
```

# Entrenamiento de la red

```python
model.fit(training_data,target_data , epochs=1000) #Aquí le ponemos lo que queremos meter y en el segundo parámetro a donde queremos llegar
```

```python
# evaluamos el modelo
scores = model.evaluate(training_data, target_data) #Aquí le preguntamos qué tan bien evalúa, le pasamos para evaluar el target data
```

```python

print (model.predict(training_data).round()) #Si le quitamos el round nos da las verdaderas salidas, ya que, redondeoa
```
