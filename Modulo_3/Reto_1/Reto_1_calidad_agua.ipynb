# Conclusiones
Se observa que se tiene una muestra de 3275 datos en total para diferentes valores de calidad química y biológica del agua, para este caso particular se ha buscado analizar la calidad del agua clasificándola en aguas duras. Esto, ya que es un problema que se presenta en diversas regiones del mundo, principalmente en aquellas donde la fuente de abstecimiento es por pozos.
De acuerdo con lo analizado, para toda la muestra se tiene un promedio de ph de 7.08 (ph neutro), una dureza promedio de 196 (agua dura), y una conductividad promedio de 426 (agua dura). 
Una vez que se aplicó el filtro para las condicones de agua dura (un ph mayor a 7.5, dureza > 180 mg/L y conductividad mayor a 300 µS/cm de acuerdo a la literatura) se obtuvo que 641 muestra de las 3275 analizadas cumplen con esta condición, teniendo un promedio de aguas duras del 19.57% sobre la muestra total; asimismo, el ph promedio de éstas es de 8.58, la dureza promedio de 214 y una conductividadpromedio de 453.
Es interesante observar que el promedio de la potabilidad ( a pesar de ser una variable dicotómica) toma valores de 0.39 (para la muestra completa) y de 0.34 para la muestra de aguas duras, es decir, la mayoria de las muestras tomadas no son potables y aptas para consumo humano.
