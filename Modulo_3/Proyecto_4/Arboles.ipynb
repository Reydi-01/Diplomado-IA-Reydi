# Implementación de árboles de decisión

```python
#Vamos a hacer un ejemplo de árbol de decisión

#Importamos librerías
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris #Importamos una base de datos de una librerías
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()
```

```python
iris
```

```python

x = iris.data
y = iris.target
```

```python
#Vamos a dividir el data set
x_train, x_test,y_train, y_test = train_test_split(x,y,test_size=0.2) #random_state=42)
```

```python
model = DecisionTreeClassifier(criterion = 'entropy').fit(x_train, y_train)
```

```python
plt.figure(figsize=(20,20))
tree.plot_tree(model, feature_names=iris.feature_names,class_names= iris.target_names, filled =True)
plt.show()
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report (y_test, y_pred, target_names=iris.target_names))
#Cuando precision y recall y f1-score es todo uno, entonces tenemos un modelo perfecto, pero sin sobre ajuste
#Cuando precision y recall son completamente diferentes  entonces hay sobre ajuste
```
