```python
#Ahora vamos a eliminar las variables que no me van a brindar información en mi modelo
valores_unicos_n = df.nunique() #Le digo que haga un conteo y lo meta en la variable en forma de serie
#De la serie de valores_unicos_n vamos a extraer los que tienen valores únicos
columnas_sin_info = valores_unicos_n[valores_unicos_n==1 ].index.tolist() #Aquí le digo que si es igual a 1, entonces me los indique, le ponemos index.list para que me devuelva sólo las variables en forma de lista y las etiqeutas
columnas_sin_info
```

```python
#Ahora vamos a eliminar esas columnas
#.drop es el comando que se usa para eliminar filas o columnas
# df.drop(columnas_sin_info, axis = 1, inplace=True)   esta es la manera para que se reemplace
df_clean= df.drop(columnas_sin_info, axis = 1) #Le pongo axis = 1 porque son columnas y inplace = True para que los reemplaze sin necesidad de reasignarlo a una variable
#La de arriba es para que se reemplace en una variable
```

```python
df_clean.head(3)
```

```python
df_clean.describe() #Obtiene un resumen estadístico, es importante resaltar que la variable describe excluye los valores nulos, por eso en el conteo no sale igual
#Se observa que python reconoce como variable numérica a las cualitativas, por ello les saca sus valores
```

## DIVISIÓN DE UN DATASET

```python
#Creamos nueva variable de las personas que siguen trabajando
siguen_trabajando = df_clean[df_clean ['Attrition' ]== 'No']
siguen_trabajando
#Co unique checamos si el data set es unico
siguen_trabajando ['Attrition' ].unique()
```

```python

#Ahora cremos un nuevo data set
renunciaron = df_clean[df_clean ['Attrition' ] =='Yes']
renunciaron
```

```python
renunciaron.describe()
```

```python
siguen_trabajando.describe()
```

EN LO  DE ARRIBA ESTAMOS VIENDO PORQUÉ RENUNCIARON O MÁS BIEN, CUÁNTOS RENUNCIARON

```python
personas_sueldo_bien = siguen_trabajando[(siguen_trabajando['MonthlyIncome' ] >10000) & (siguen_trabajando['MonthlyIncome' ] <15000) ] #ESTA CONDICIÓN NOS LIMITA A LAS PERSONAS QUE GANAN ENTRE 10000 O 150000
personas_sueldo_bien.describe()
```

```python
personas_sueldo_bien_distancia = siguen_trabajando[(siguen_trabajando['MonthlyIncome' ] >10000) & (siguen_trabajando['DistanceFromHome' ] >8) ] #Filtro para los que ganen más de 10000 y viven a mas de 8 km. Operador & (and)
personas_sueldo_bien_distancia.describe()
```

```python
#También puedp filtrar para las personas que ganen mas de 10000 o vivan a 8 km
personas_sueldo_bien_odistancia = siguen_trabajando[(siguen_trabajando['MonthlyIncome' ] >10000) | (siguen_trabajando['DistanceFromHome' ] >8) ] #Filtro para los que ganen más de 10000 o viven a mas de 8 km, el operador es | (operador or)
personas_sueldo_bien_odistancia.describe()
```

```python
personas_sueldo_bien_ydistancia_ysatisfecho = siguen_trabajando[(siguen_trabajando['MonthlyIncome' ] >10000) & (siguen_trabajando['EnvironmentSatisfaction' ] ==4) & (siguen_trabajando[ 'DistanceFromHome']>8) ] #Filtro para los que ganen más de 10000 y estén satisfechos con el trabajo y viven a 8 km
personas_sueldo_bien_ydistancia_ysatisfecho.describe()
```

```python
personas_sueldo_bien_ydistancia_ysatisfecho.head()
```

```python
#Si yo quiero ver qué rol cumplen los que filtré arriba
personas_sueldo_bien_ydistancia_ysatisfecho['JobRole' ].unique() #Recordar que unique me indica valores únicos
```

```python
personas_sueldo_bien_ydistancia_ysatisfecho[ personas_sueldo_bien_ydistancia_ysatisfecho['JobRole' ]=='Sales Executive'].describe()
#Aquí quiero que me diga con el describe cuántos tiene el cargo de Sales Executive y en promedio cuánto ganan
```
