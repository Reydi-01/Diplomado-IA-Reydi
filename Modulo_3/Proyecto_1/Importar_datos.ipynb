## Detección de valores atípicos

```python
#vamos a importar libreria de gráficos
import matplotlib.pyplot as plt
```

```python
#vamos a hacer un histograma de nuestros datos para la columna edad
plt.hist(df_clean['Age' ], bins =50)
plt.show()
#Un metodo sencillo para ver si hay datos atípicos es graficarlos como lo hicimos y si no se observa algún valor o patrón diferente, vamos bien
```

```python
#Importamos librería sanbors
import seaborn as sns
#Me permite hacer gráficos más bonitos y de forma sencilla
sns.distplot(x=df_clean['Age' ])
plt.show()
```

```python
#Otro gráfico que sirve mejor para esto es el de cajas y bigotes
df_clean.boxplot(column='Age') #Pandas me permite graficar
```

```python
#Para calcular los cuantiles
Q1 = df_clean['Age' ].quantile(0.25)
Q3= df_clean['Age' ].quantile(0.75)
IQR = Q3-Q1 #PARA CALCULAR EL RANGO INTERCUÁNTILICO

limite_inferior = Q1 - 1.5*IQR
limite_inferior #Si la edad es menor a esto que se calcula, entonces se considera dato atípico
#Ahora hacemos lo de arriba y si es mayor es atípico
limite_superior = Q3 + 1.5*IQR
limite_superior
```

## Simulación de datos atípicos

```python
import numpy as np
df_test = df_clean.copy () #Hacemos una copia de mis datos limpios
#Ahora generaremos datos atípico
indices = np.random.choice(df_test.index, 2, replace=False)
df_test.loc[indices[0 ], 'Age' ]+= 3 * IQR
df_test.loc[indices[1 ], 'Age' ]+= 3 * IQR
```

```python
df_test.boxplot(column='Age')
```

```python
#Para volver a calcular los límite de los datos atípicos
#Para calcular los cuantiles
Q1 = df_test['Age' ].quantile(0.25)
Q3= df_test['Age' ].quantile(0.75)
IQR = Q3-Q1 #PARA CALCULAR EL RANGO INTERCUÁNTILICO

limite_inferior = Q1 - 1.5*IQR
limite_inferior #Si la edad es menor a esto que se calcula, entonces se considera dato atípico
#Ahora hacemos lo de arriba y si es mayor es atípico
limite_superior = Q3 + 1.5*IQR
limite_superior
```
