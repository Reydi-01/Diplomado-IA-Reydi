```python
experiencia = [ 1,3,5,7,9]
salario_data_scientists = [3434, 6436, 2343, 4656, 8999 ]
salario_ing_software = [3434, 63446, 2343, 46456, 84999 ]
```

```python
#Si yo quiero poner dos gráficas que tienen relación entre sí
plt.plot(experiencia, salario_data_scientists )
plt.plot(experiencia, salario_ing_software)
plt.title('Salarios')
plt.xlabel('Experiencia')
plt.ylabel('Ganancias')
plt.show
```

```python
#Si yo quiero que se agreguen las leyendas
plt.plot(experiencia, salario_data_scientists )
plt.plot(experiencia, salario_ing_software)
plt.title('Salarios')
plt.xlabel('Experiencia')
plt.ylabel('Ganancias')
plt.legend([ 'Data scientifist', 'Ing software' ])
plt.show
```

```python
#Si yo quiero modificar el color de mis gráficas y además las etiquetas se las puedo poner así en los títulos
plt.plot(experiencia, salario_data_scientists, color = 'r', label = 'Data scientist'  )
plt.plot(experiencia, salario_ing_software, color = 'g', label = 'Ing Software')
plt.title('Salarios')
plt.xlabel('Experiencia')
plt.ylabel('Ganancias')
plt.legend()
plt.show
```

```python
#También puede ser color hexagecimal
#Si yo quiero modificar el color de mis gráficas y además las etiquetas se las puedo poner así en los títulos
plt.plot(experiencia, salario_data_scientists, color = '#628c61', label = 'Data scientist'  )
plt.plot(experiencia, salario_ing_software, color = '#0f0f0f', label = 'Ing Software')
plt.title('Salarios')
plt.xlabel('Experiencia')
plt.ylabel('Ganancias')
plt.legend()
plt.show
```

```python
plt.plot(experiencia, salario_data_scientists, color = '#628c61', label = 'Data scientist', linestyle = '--', marker = 'o'  ) #Agregr estilo de linea y marcador
plt.plot(experiencia, salario_ing_software, color = '#0f0f0f', label = 'Ing Software', linestyle = ':', marker = '*') #Agregar estilo de línea y marcador
plt.title('Salarios')
plt.xlabel('Experiencia')
plt.ylabel('Ganancias')
plt.legend()
plt.grid (True) #Para agregar malla de datos
plt.show
```

```python
#Vamos a hacer gráfico de dispersión
import numpy as np
```

```python
#Vamos a ocupar numpy para generar datos aletorios para graficar gráfico de dispersión
x = np.random.rand(100)
y = np.random.rand(100)
```

```python
plt.scatter(x,y, color = 'r', alpha = 0.6) #ÉSTE ES EL COMANDO PARA EL GRÁFICO DE DISPERSIÓN, color es para el color y alpha para transparencia
plt.title('Gráfico de dispersión')
plt.xlabel('X disperso')
plt.ylabel('Y disperso') 
plt.show
```

```python
#Podemos hacer histogramas
datos = np.random.randn(1000) #Genero valores pseudo aleatorios con distribución normal
plt.hist(datos, bins = 50) #bins se refiere a los intervlos
plt.title('Histograma')
plt.xlabel('Valor')
plt.ylabel('Frecuencia')
plt.show
```

```python
#Podemos generar un gráfico de cajas y bigotes
#Vamos a generar datos ramdom con distribución normal
datos = [np.random.normal(0,std,100) for std in range (1,4) ]
plt.boxplot(datos, vert=True) #Aquí en vert le puedo poner False y me las pone horizontal
plt.show()
```

```python
#Vamos a crear series de tiempo
#Importamos pandas
import pandas as pd
data_range = pd.date_range(start= '2023-01-01', end = '2023-12-31', freq = 'D') #Esto es para generar el rango de datos que es de un año y la frecuencia es de un día (D)
#Vamos a generar nuestros datos
precios = np.cumsum(np.random.randn(len(data_range))) + 100 
```

```python
#Se lo asignamos a una variable y lo convertimos a serie
data = pd.DataFrame({'Date': data_range, 'Precios': precios})
data.head()
```

```python
#Con los datos de arriba genero las series de tiempo que son las fechas contra el precio
plt.plot(data['Date' ], data['Precios' ], label = 'Precio')
plt.show()
#Una serie de tiempo la puedo usar para modelar fenómenos donde la variable independiente sea el tiempo
```

```python
#Generamos gráfico de senos y cosenos
x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)
#Geenramos esto para la gráfica de barras
categorias = [ 'a', 'b', 'c', 'd']
valores = [5,7,3,8]
```

## Reto2: Generar gráfico de cuatro gráficos

```python
#Vamos a generar una cuadrícula de figuras
fig, axs = plt.subplots(2,2, figsize = (15,10)) #El 2,2 me indica las dimensiones que tendré
axs [0,0].plot(x, y1, label = 'senos')
axs [0,0].plot(x, y2, label = 'cosenos')
axs [0,0].legend(loc='upper left') #Aquí el upper me dice donde lo quiero, rigth es derecha o bien left para izquierda
axs [0,0].set_title('Gráfico de lineas')
axs[0,1].bar(categorias, valores)
axs [0,1].set_title('Gráfico de barras') #Para ponerle título a cada
axs[1,0].hist(datos, bins = 50)
axs[1,0 ].set_title('Histograma')
axs [1,1].plot(data['Date' ], data['Precios' ], label = 'Precio')
axs[1,1].set_title('Serie de tiempo')
#Para poner título al gráfico
fig.suptitle('Ejemplos de gráficos')
plt.show()
```
